$$
 p_{t} = p_0
        +  \rho_{p} p_{t}

## White Standard Deviation

I numerically calculated White Standard Deviation for each parameters.

$$
\text{Cov}(\hat{\theta}) = n\times H(\hat{\theta})^{-1} S(\hat{\theta}) H(\hat{\theta})^{-1}
$$
where n is the time series length.

The Hessian matrix $H(\theta)$ is the matrix of second-order partial derivatives of the log-likelihood function with respect to the parameters:

$$
H(\theta) = \frac{\partial^2 \log L(\theta)}{\partial \theta \partial \theta'}
$$

I evaluate this at the estimated parameters $\hat{\theta}$, where $\log L(\theta) = \sum_t \log L_t(\theta) $ is the sum of log likelihood of all points.

The score vector is the first derivative of the log-likelihood function with respect to the parameters:

$$
s_t(\theta) = \frac{\partial \log L_t(\theta)}{\partial \theta}
$$

where $\log L_t(\theta)$ is the contribution to the log-likelihood function from observation $t$.

The outer product of the score vectors is:

$$
S(\theta) = \sum_{t=1}^T s_t(\theta) s_t(\theta)'
$$


### The reason of very small standard deviations and very large t values.

Sometimes the t values of some parameters are very big and optimizer stops iteration. In my opinion, this is an indicator of local optima. When the optimizer encounters a point where the slopes in certain parameter dimensions are extremely steep or approaches an asymptote, it results in very small standard deviations and disproportionately large t-values for that parameters.
